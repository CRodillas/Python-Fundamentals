```python
import os
import re
import json
from collections import defaultdict, Counter
```

```python
def load_data_from_json(filepath = '/Users/christianmiguelrodillas/Downloads/emails_data.json'):

    try:
        with open(filepath, 'r') as file:
            data = json.load(file)
        return [(item['text'], item['label']) for item in data]
    except FileNotFoundError:
        print(f"File not found: {filepath} not")
```

```python
emails = load_data_from_json()

emails
```

this verifies that the json exists

```python
spam_words_counts = Counter()
ham_words_counts = Counter()

total_spam_docs = 0
total_ham_docs = 0
```

```python
if not emails:
    print("No emails loaded. Please check the file path and ensure the JSON file exists.")
    exit()
```

```python
for text, labels in emails:
    words = re.sub(r'[^\w\s]', '', text.lower()).split()
    if labels == 'spam':
        spam_words_counts.update(words)
        total_spam_docs += 1
    else:
        ham_words_counts.update(words)
        total_ham_docs += 1
```

```python
vocab = list(spam_words_counts.keys() | ham_words_counts.keys())
total_spam_words = sum(spam_words_counts.values())
total_ham_words = sum(ham_words_counts.values())
smoothing = 1


word_probs_spam = defaultdict(lambda: smoothing / (total_spam_words + len(vocab) * smoothing))
for word, count in spam_words_counts.items():
    word_probs_spam[word] = (count + smoothing) / (total_spam_words + len(vocab) * smoothing)   

word_probs_ham = defaultdict(lambda: smoothing / (total_ham_words + len(vocab) * smoothing))
for word, count in ham_words_counts.items():
    word_probs_ham[word] = (count + smoothing) / (total_ham_words + len(vocab) * smoothing)
```

s4: predict

```python
new_message = input("Enter a message to classify: ")
new_words = re.sub(r'[^\w\s]', '', new_message).lower().split()


    #calculate probabilities
prob_spam = total_spam_docs / (total_spam_docs + total_ham_docs)
prob_ham = total_ham_docs / (total_spam_docs + total_ham_docs)


for word in new_words:
    prob_spam *= word_probs_spam[word]
    prob_ham *= word_probs_ham[word]


print(f"Message {new_message}")
print(f"Probability of being spam: {prob_spam}")
print(f"Probability of being ham: {prob_ham}")
print("Classified as:", "spam" if prob_spam > prob_ham else "ham")
```
